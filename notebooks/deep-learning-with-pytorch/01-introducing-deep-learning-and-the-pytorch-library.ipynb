# Chapter 1: Introducing Deep Learning and the PyTorch Library

This notebook contains the interactive code examples for **Chapter 1** from *Deep Learning with PyTorch (2nd Edition)*.

## 1. Verifying System and PyTorch Installation

```python
import sys
import torch

print('=== System and PyTorch Diagnostics ===')
print('Python Version:', sys.version.split()[0])
print('PyTorch Version:', torch.__version__)

cuda_available = torch.cuda.is_available()
print('CUDA Available:', cuda_available)

if cuda_available:
    device_count = torch.cuda.device_count()
    device_name = torch.cuda.get_device_name(0)
    print('Number of GPUs:', device_count)
    print('Primary GPU Device Name:', device_name)
else:
    print('Running on CPU mode.')
```

## 2. Creating and Inspecting Tensors (0D to 3D)

```python
# 1. Scalar (0-dimensional tensor)
scalar = torch.tensor(42.0)
print('Scalar:', scalar)
print('Scalar dimension (ndim):', scalar.ndim)

# 2. Vector (1-dimensional tensor)
vector = torch.tensor([1.5, 3.0, 4.5])
print('\nVector:', vector)
print('Vector shape:', vector.shape)

# 3. Matrix (2-dimensional tensor: 2 rows, 3 columns)
matrix = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
print('\nMatrix:\n', matrix)
print('Matrix shape:', matrix.shape)

# 4. 3D Tensor representing a small RGB image (3 channels x 2 height x 2 width)
rgb_image = torch.zeros((3, 2, 2))
print('\n3D Tensor shape:', rgb_image.shape)
```

## 3. Dynamic Control Flow in PyTorch (Eager Mode)

```python
x = torch.tensor([2.0, -3.0, 5.0])

for val in x:
    if val > 0:
        print('Positive value:', val.item())
    else:
        print('Negative value:', val.item())
```

## 4. Accelerated Matrix Multiplication on GPU/CPU

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('Using device:', device)

a = torch.randn(1000, 1000, device=device)
b = torch.randn(1000, 1000, device=device)
c = torch.matmul(a, b)

print('Matrix multiplication output shape:', c.shape)
```
